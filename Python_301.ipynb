**Part 1 - Load and understand the data**

Importing pandas and numpy

```python
import pandas as pd
import numpy as np
```

Loading data directly from web so that nothing loses when runtime restarts

```python
print(df.shape)
print(f"Number of rows: {df.shape[0]}")
print(f"Number of columns: {df.shape[1]}")
```

Listing column names

```python
print(df.columns)
```

Checking data types of each column

Counting missing values in every column

Counting the duplicated rows

Description and CustomerID have null values , so they need cleaning

**Part 2- Clean the data**

```python
#copy of clean original data
clean_df = df.copy()
print(clean_df.shape)
```

Keeping missing customerID becuase dropping them could impact overall data analysis as they are large in number

Checking data type of InvoiceDate

```python
clean_df['InvoiceDate'].dtype
```

Converting it to proper date time column from text

```python
pd.to_datetime(clean_df['InvoiceDate'])
```

Dropping exact duplicates

```python
clean_df = clean_df.drop_duplicates()
print(clean_df.shape)
```

Removing invalid transactions

```python
clean_df = clean_df[(clean_df['Quantity'] > 0) & (clean_df['UnitPrice'] > 0)]
print(clean_df.shape)
```

Creating a new sales column

```python
clean_df['Sales'] = clean_df['Quantity'] * clean_df['UnitPrice']
print(clean_df.shape)
```

**Part 3- NumPy analysis**

```python
print(f"mean of sales: {np.mean(clean_df['Sales']):.2f}")
print(f"median of sales: {np.median(clean_df['Sales']):.2f}")
print(f"minimum of sales:{np.min(clean_df['Sales']):.2f}")
print(f"maximum of sales: {np.max(clean_df['Sales']):.2f}")
print(f"standard deviation of sales:{np.std(clean_df['Sales']):.2f}")
```

mean is almost twice of median , which tells us that sales are right skewed due to high value outliers

**Part 4- Pandas filtering**
