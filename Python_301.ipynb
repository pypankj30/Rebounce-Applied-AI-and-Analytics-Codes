**Part 1 - Load and understand the data**

Importing pandas and numpy

```python
import pandas as pd
import numpy as np
```

Loading data directly from web so that nothing loses when runtime restarts

```python
url = "https://huggingface.co/datasets/aarav912/online-retail/resolve/main/online_retail.csv"
df = pd.read_csv(url)
```

```python
print(df.shape)
print(f"Number of rows: {df.shape[0]}")
print(f"Number of columns: {df.shape[1]}")
```

Listing column names

```python
print(df.columns)
```

Checking data types of each column

```python
df.info()
```

Counting missing values in every column

```python
df.isna().sum()
```

Counting the duplicated rows

```python
df.duplicated().sum()
```

Description and CustomerID have null values , so they need cleaning

**Part 2- Clean the data**

```python
#copy of clean original data
clean_df = df.copy()
print(clean_df.shape)
```

Keeping missing customerID becuase dropping them could impact overall data analysis as they are large in number

Checking data type of InvoiceDate

```python
clean_df['InvoiceDate'].dtype
```

Converting it to proper date time column from text

```python
pd.to_datetime(clean_df['InvoiceDate'])
```

Dropping exact duplicates

```python
clean_df = clean_df.drop_duplicates()
print(clean_df.shape)
```

Removing invalid transactions

```python
clean_df = clean_df[(clean_df['Quantity'] > 0) & (clean_df['UnitPrice'] > 0)]
print(clean_df.shape)
```

Creating a new sales column

```python
clean_df['Sales'] = clean_df['Quantity'] * clean_df['UnitPrice']
print(clean_df.shape)
```

**Part 3- NumPy analysis**

```python
print(f"mean of sales: {np.mean(clean_df['Sales']):.2f}")
print(f"median of sales: {np.median(clean_df['Sales']):.2f}")
print(f"minimum of sales:{np.min(clean_df['Sales']):.2f}")
print(f"maximum of sales: {np.max(clean_df['Sales']):.2f}")
print(f"standard deviation of sales:{np.std(clean_df['Sales']):.2f}")
```

mean is almost twice of median , which tells us that sales are right skewed due to high value outliers

**Part 4- Pandas filtering**

```python
print(f"Number of transactions having sales above 500: {clean_df[clean_df['Sales'] > 500].shape[0]}")
```

**Part 5- GroupBy and aggregation**

```python
summary_table = clean_df.groupby('Country').agg(
    Num_transactions=('InvoiceNo', 'count'),
    Total_sales=('Sales', 'sum'),
    Average_sales=('Sales', 'mean')
).reset_index()
print(summary_table)
```

which 5 countries have highest total sales

```python
print(f"5 countries with the highest total sales:")
summary_table.sort_values('Total_sales', ascending=False).head(5)
```

Country with highest average transaction value

```python
print(f"Country with the highest average transaction value:")
summary_table.sort_values('Average_sales', ascending=False).head(1)
```

Number of transactions of top selling country

```python
print(f"Number of transactions of top selling country:")
summary_table.sort_values('Num_transactions', ascending=False).head(1)
```

**Part 6- Merge and region analysis**

Turning dictionary into small dataframe

```python
region_lookup = {
"United Kingdom": "Europe",
"France": "Europe",
"Germany": "Europe",
"Spain": "Europe",
"Netherlands": "Europe",
"Belgium": "Europe",
"Switzerland": "Europe",
"Portugal": "Europe",
"Australia": "Oceania",
"Japan": "Asia",
"USA": "Americas",
}
region_lookup_df=pd.DataFrame(list(region_lookup.items()), columns=['Country', 'Region'])
region_lookup_df.head()
```

Merging into country summary table

```python
country_summary_table = pd.merge(summary_table, region_lookup_df, on='Country', how='left')
country_summary_table['Region'] = country_summary_table['Region'].fillna('Other')
country_summary_table.head()
```

Grouping by region and finding total sales

```python
summary_region = country_summary_table.groupby('Region')['Total_sales'].sum().reset_index()
print(f"Total sales by region:")
summary_region

```

Region with highest total sales

```python
print("Region with the highest total sales:")
summary_region.sort_values('Total_sales', ascending=False).head(1)
```
