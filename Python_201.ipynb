**A . Personal Expense Tracker**
It is a program that will track expenses , will help identify the pattern of expenses , categories , savings .

```python
import json
import os
```

**Step 1 . Adding new expenses to the list**

```python
expenses = []
expense1 = {"category":"food", "amount":250, "description":"lunch"}
expense2 = {"category":"vegetabes", "amount":150, "description":"curry"}
expense3 = {"category":"nutrition", "amount":300, "description":"protein"}
expense4 = {"category":"exercise", "amount":200, "description":"gym"}
expense5 = {"category":"travel", "amount":200, "description":"morning"}
expense6 = {"category":"entertainment", "amount":500, "description":"movie"}
expense7 = {"category":"shopping", "amount":200, "description":"daily needs"}
expense8 = {"category":"food", "amount":300, "description":"dinner"}
expense9 = {"category":"wifi", "amount":30, "description":"internet"}
expense10 = {"category":"travel", "amount":200, "description":"evening"}

expenses.append(expense1)
expenses.append(expense2)
expenses.append(expense3)
expenses.append(expense4)
expenses.append(expense5)
expenses.append(expense6)
expenses.append(expense7)
expenses.append(expense8)
expenses.append(expense9)
expenses.append(expense10)

display(expenses)
```

**Step 2 . Creating function to view expenses**

```python
def view_expenses():
    for expense in expenses:
        print(f"""Category: {expense['category']}, Amount: {expense['amount']}, Description: {expense['description']}""")
```

### Step 3. Creating a function to view expenses by category

```python
def view_expenses_by_category(category):
    print(f"Viewing expenses for category: {category}")
    found_expenses = False
    for expense in expenses:
        if expense["category"] == category:
            print(f"""Category: {expense['category']}, Amount: {expense['amount']}, Description: {expense['description']}""")
            found_expenses = True
    if not found_expenses:
        print(f"No expenses found for category: {category}")
```

Now let's view only the 'food' category expenses.

```python
view_expenses_by_category("travel")
```

```python
view_expenses()
```

**Step 4. Creating a function to view total amount spent**


```python
def calculate_total_amount_spent():
    total_amount = 0
    for expense in expenses:
        total_amount += expense["amount"]
    print(f"Total amount spent: {total_amount}")
```

```python
calculate_total_amount_spent()
```

**Step 5. Creating function to view totals by category**

```python
def calculate_total_amount_by_category(category):
    total_amount = 0
    for expense in expenses:
        if expense["category"] == category:
            total_amount += expense["amount"]
    return total_amount
```

```python
print(calculate_total_amount_by_category("food"))
```
