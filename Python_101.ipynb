**Problem 5. The Password Strength Check**

```python
password = "password@3020"
password_length_ok = len(password) >= 8
password_contains = "@" in password
is_strong = password_length_ok and password_contains

print(f"Password: {password}")
print(f"Length OK: {password_length_ok}")
print(f"Contains '@': {password_contains}")
print(f"Is strong: {is_strong}")
```

**Problem 6 . The Class Roster**

**Problem 8. Squares of the even Numbers**

```python
range(1,21)
even_squares = [x**2 for x in range(1,21) if x % 2 == 0]
print(even_squares)
```

**Problem 9. Sort Products by price**

```python
products =[
    ("Notebook",250),
    ("Backpack",1800),
    ("Pen",20),
    ("Eraser",15),
    ("Water Bottle",350)
]
products = sorted(products, key=lambda x: x[1])
print(products)
```

**Problem 10 . Filter the Passing students**
